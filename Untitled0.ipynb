```python
import numpy as np
import matplotlib.pyplot as plt
import os
import urllib.request
import cv2
```

IMPORTING NAD LOADING MNIST **DATASET**

```python
url = "https://storage.googleapis.com/tensorflow/tf-keras-datasets/mnist.npz"

urllib.request.urlretrieve(url, "mnist.npz")

print("Dataset downloaded successfully!")
print("File size:", round(os.path.getsize("mnist.npz") / (1024 * 1024), 2), "MB")
```

DISPLAY DATASET **DIMENSION**

```python
with np.load("mnist.npz") as data:
    x_train = data["x_train"]
    y_train = data["y_train"]
    x_test = data["x_test"]
    y_test = data["y_test"]

print("Training images:", x_train.shape)
print("Training labels:", y_train.shape)
print("Testing images:", x_test.shape)
print("Testing labels:", y_test.shape)
```

**NORMALISATION**

```python

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0


print("Training pixel range:", x_train.min(), "to", x_train.max())
print("Testing pixel range:", x_test.min(), "to", x_test.max())
```

IMAGE **PREPROCESSING**

```python
img = x_train[0].squeeze()
img = (img * 255).astype(np.uint8)


average_blur = cv2.blur(img, (3, 3))
gaussian_blur = cv2.GaussianBlur(img, (3, 3), 0)
median_blur = cv2.medianBlur(img, 3)


kernel = np.array([
    [0, -1, 0],
    [-1, 5, -1],
    [0, -1, 0]
])
sharpened = cv2.filter2D(img, -1, kernel)


sobel_x = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)

sobel = cv2.convertScaleAbs(
    cv2.magnitude(sobel_x, sobel_y)
)

laplacian = cv2.convertScaleAbs(
    cv2.Laplacian(img, cv2.CV_64F)
)

canny = cv2.Canny(img, 50, 150)


h, w = img.shape
rotation_matrix = cv2.getRotationMatrix2D(
    (w / 2, h / 2), 30, 1.0
)
rotated = cv2.warpAffine(img, rotation_matrix, (w, h))


images = [
    ("Original", img),
    ("Average Blur", average_blur),
    ("Gaussian Blur", gaussian_blur),
    ("Median Blur", median_blur),
    ("Sharpened", sharpened),
    ("Sobel Edges", sobel),
    ("Laplacian Edges", laplacian),
    ("Canny Edges", canny),
    ("Rotated 30 Degrees", rotated)
]

fig, axes = plt.subplots(3, 3, figsize=(9, 10))

for ax, (title, image) in zip(axes.flat, images):
    ax.imshow(image, cmap="gray", vmin=0, vmax=255)
    ax.set_title(title)
    ax.axis("off")

plt.suptitle(
    f"MNIST Image Preprocessing — Label: {y_train[0]}",
    fontsize=14
)
plt.tight_layout()
plt.show()
```
